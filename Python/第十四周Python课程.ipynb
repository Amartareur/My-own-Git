```python
#1.（1）定义一个 Animal 父类，包含 name 属性和方法 speak()（打印“动物叫了一声”）。
#（2）定义 Dog 子类继承 Animal，使用 super() 调用父类构造方法，重写 speak() 方法打印“小狗汪汪叫”。
#（3）创建 Dog 对象并调用 speak() 方法。
class Animal:
    def __init__(self,name):
        self.name = name
        
    def speak(self):
        print("动物叫了一声")

class Dog(Animal):
    def __init__(self, name):
        super().__init__(name)

    def speak(self):
        print("{},小狗汪汪叫".format(self.name))

dog1 = Dog("lucky")
dog1.speak()
```

```python
#2.  定义一个 Vector3D 类，包含（1）属性：x、y、z（三维坐标）；
#（2）__add__：两个向量相加；
#（3）__sub__：两个向量相减；
#（4）__mul__：向量与标量相乘；
#（5）__str__：返回向量字符串，格式为 (x, y, z).
#创建两个向量，测试加、减、乘运算。
class Vector3D:
    def __init__(self,x,y,z):
        self.x = x
        self.y = y
        self.z = z
        
    def __add__(self,other):
        return (self.x + other.x, self.y + other.y, self.z + other.z)

    def __sub__(self,other):
        return (self.x - other.x, self.y - other.y, self.z - other.z)

    def __mul__(self,other):
        return (self.x * other.x, self.y * other.y, self.z * other.z)
    
    def __str__(self,other):
        print("第一个向量是（{}，{}，{}），第二个向量是（{}，{}，{}）".format(self.x,self.y,self.z,other.x,other.y,other.z))

Vector1 = Vector3D(2,7,4)
Vector2 = Vector3D(3,2,4)
Vector1.__add__(Vector2)
```

```python
Vector1.__str__(Vector2)
```

```python
#4. 综合题目：银行账户系统
#银行账户有不同的类型，每种类型的取款规则不同。请设计一个账户类族，通过继承和多态管理不同类型的账户。
#要求：
#（1）定义父类 BankAccount：
#属性：owner（账户名）、balance（余额）
#方法 deposit(amount)：存款，余额增加
# withdraw(amount)：取款，余额减少（如果余额不足返回“余额不足”）
#方法 show()：显示账户信息
#（2）定义子类 SavingsAccount（储蓄账户）继承 BankAccount：
#增加属性：rate（年利率，如0.03表示3%）
#方法 add_interest()：计算利息并存入
#重写 show()：调用父类 show() 后补充利率信息
#（3）定义子类 CreditAccount（信用账户）继承 BankAccount：
#增加属性：limit（信用额度）
#重写 withdraw()：余额不足时可透支，但不能超过额度
#重写 show()：调用父类 show() 后补充额度信息
#（4）创建不同类型的账户对象，测试各种操作。
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        return f"存款 {amount} 元，当前余额 {self.balance} 元"

    def withdraw(self, amount):
        if self.balance >= amount:
            self.balance -= amount
            return f"取款 {amount} 元，当前余额 {self.balance} 元"
        else:
            return "余额不足"

    def show(self):
        return f"余额: {self.balance} 元"


class SavingsAccount(BankAccount):
    def __init__(self, owner, balance, rate):
        super().__init__(owner, balance)
        self.rate = rate

    def add_interest(self):
        interest = self.balance * self.rate
        self.balance += interest
        return f"计算利息成功！结算利息: {interest} 元，当前余额为: {self.balance} 元"

    def show(self):
        base_info = super().show()
        return f"{base_info}, 年利率: {self.rate * 100}%"


class CreditAccount(BankAccount):
    def __init__(self, owner, balance, limit):
        super().__init__(owner, balance)
        self.limit = limit  

    def withdraw(self, amount):
        if self.balance + self.limit >= amount:
            self.balance -= amount
            return f"取款 {amount} 元 成功（含透支），当前余额 {self.balance} 元"
        else:
            return f"取款失败：超过信用额度！最高可透支额度为 {self.limit} 元"

    def show(self):
        base_info = super().show()
        return f"{base_info}, 信用额度: {self.limit} 元"

print("====== 1. 测试基础账户 ======")
an1 = BankAccount("Amartareur", 10000)
print(an1.deposit(500))
print(an1.withdraw(1000))
print(an1.withdraw(15000))  # 测试余额不足
print(an1.show())

print("\n====== 2. 测试储蓄账户 ======")
an2 = SavingsAccount("Amartareur", 10000, 0.03)
print(an2.show())
print(an2.add_interest())
print(an2.show())

print("\n====== 3. 测试信用账户 ======")
an3 = CreditAccount("Amartareur", 1000, 5000) 
print(an3.show())
print(an3.withdraw(3000))
print(an3.show())
print(an3.withdraw(4000))
print(an3.deposit(2000))
print(an3.show())
```
