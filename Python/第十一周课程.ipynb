```python
def poetory(title, author, **details):
    print(f"标题：{title}")
    print(f"作者：{author}")
    print(f"年份：{details.get('year', '未知')}")

# 调用示例
poetory("静夜思", "李白", year="唐代")
poetory("春江花月夜", "张若虚", year="唐代")
```

```python
#1. 编写一个函数 circle_area(r)，参数 r 为圆的半径，返回圆的面积（π取3.14）。调用该函数，分别计算半径为3、5、7的圆面积并输出。
def circle_area(r):
    pi = 3.14 #定义pi的大小
    area = pi * r ** 2 #定义计算公式
    return area #返回并输出

print(circle_area(3))
print(circle_area(5))
print(circle_area(7))
```

```python
#2. 编写一个函数 student_info(name, age, major)，打印学生信息。分别使用位置参数和关键字参数两种方式调用该函数，输出自己的姓名、年龄和专业。
def student_info(name, age, major):
    print(f"名字：{name} 年龄：{age} 专业：{major}")#打印三个内容

student_info("蒋博文", 19, "数字媒体技术")
```

```python
#3. 编写一个函数 greet(name, greeting="你好")，greeting 参数有默认值“你好”。分别尝试：（1）只传入姓名，使用默认问候语（2）传入姓名和自定义问候语
#调用函数输出结果。
def greet(name, greeting="你好"):
    print(f"{greeting}，{name}")#就是打印两个内容而已

greet("Bob")
greet("Jack", "Hello")
```

```python
#4. 编写一个函数 average(*args)，可以接收任意多个数值，计算并返回它们的平均值。分别测试传入3个数和5个数的情况。
def average(*args):
    count = len(args)  # 统计个数
    total = sum(args)  # 计算总和
    return total / count

average(4,6,8)
```

```python
average(4,6,8,10,12)
```

```python
#5. 诗词信息整理函数
#毛泽东同志不仅是伟大的无产阶级革命家，也是一位气势磅礴的诗人。他的诗词记录了中国革命的光辉历程，展现了共产党人的豪情壮志与革命英雄主义气概。请编写以下函数：
#（1）定义函数
#poem_info(title, author, dynasty=“近现代”, **kwargs)，其中：title 和 author 为必填参数；dynasty 有默认值“近现代”；
#kwargs 接收其他任意键值对信息（如名句、创作时间、创作地点等）。

#（2）已有诗词信息如下（从中任选两首，调用函数整理并打印）：
#《七律·长征》：作者毛泽东，名句“红军不怕远征难，万水千山只等闲”，创作于1935年10月；
#《沁园春·雪》：作者毛泽东，名句“数风流人物，还看今朝”，创作于1936年2月；
#《沁园春·长沙》：作者毛泽东，名句“问苍茫大地，谁主沉浮”，创作于1925年秋，创作于湘江橘子洲头；
def poem_info(title, author, dynasty="近现代", **kwargs):
    print("=====诗词信息=====")
    print("标题：", title)
    print("作者：", author)
    print("朝代：", dynasty)
    for key, value in kwargs.items():
        print(f"{key}：{value}")
    print("=================")
    
poem_info("七律·长征", "毛泽东", 名句="红军不怕远征难，万水千山只等闲", 创作时间="1935年10月")
poem_info("沁园春·雪", "毛泽东", 名句="数风流人物，还看今朝", 创作时间="1936年2月")
```
