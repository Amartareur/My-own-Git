```python
#1.定义一个 Circle 类，表示一个圆，包含以下要求（1）使用 __init__() 构造方法初始化半径 radius；
#（2）方法 area()：计算并返回圆的面积（π取3.14）；
#（3）方法 perimeter()：计算并返回圆的周长（π取3.14）。
#创建两个圆对象（半径分别为3和5），分别调用 area() 和 perimeter() 方法。
#输出示例：半径为3的圆：面积 = 28.26，周长 = 18.84
#半径为5的圆：面积 = 78.5，周长 = 31.4
class Circle:
    pi = 3.14
    def __init__(self,radius):
        self.radius = radius
        print("半径为{}的圆".format(self.radius))
    def area(self):
        S = self.pi * self.radius**2
        print("面积 = {}".format(S))
    def perimeter(self):
        D = 2 * self.pi * self.radius
        print("周长 = {}".format(D))

a = Circle(3)
a.area()
a.perimeter()
```

```python
a = Circle(5)
a.area()
a.perimeter()
```

```python
#2.定义一个 Book 类，包含以下要求（1）使用 __init__() 构造方法初始化书名 title、作者 author、价格 price，出版社 press 有默认值“清华大学出版社”；
#（2）方法 info()：打印图书完整信息；
#（3）方法 discount(percent)：按折扣计算折后价格并返回（percent 为折扣数，如8表示8折）。
#创建两本书对象（一本使用默认出版社，一本指定其他出版社），分别调用 info() 和 discount() 方法。
#输出示例： 《Python程序设计》  作者：董付国  出版社：清华大学出版社  定价：59.0元 打8折后：47.2元
#《Python编程》  作者：Eric Matthes  出版社：人民邮电出版社  定价：89.0元 8打7折后：62.3元
class Book:
    def __init__(self, title, author, price, press="清华大学出版社"):
        self.title = title
        self.author = author
        self.price = price
        self.press = press
    def info(self):
        print("《{}》 作者：{}  出版社：{} 定价：{}元  ".format(self.title,self.author,self.press,self.price))
    def discount(self,percent):
        the_percent = percent/10 * self.price
        print("打{}折后：{}".format(percent,the_percent))

a = Book('Python程序设计','董付国',59)
a.info()
a.discount(8)
```

```python
#3.定义一个 Employee 类，包含以下要求：（1）类属性 company = “北京印刷学院”；
#（2）实例属性：name（姓名）、position（职位）、salary（月薪）；
#（3）方法 show()：打印员工信息；（4）方法 annual_salary()：计算并返回年薪（月薪 × 12）
#创建两个员工对象，分别打印它们的类属性和实例属性，然后修改类属性，再次打印。
#输出示例：姓名：张三，职位：教师，月薪：8000元，单位：北京印刷学院
#姓名：李四，职位：辅导员，月薪：6000元，单位：北京印刷学院
#张三的年薪：96000元
#李四的年薪：72000元
#修改类属性后——
#姓名：张三，职位：教师，月薪：8000元，单位：北京印刷学院新媒体学院
#姓名：李四，职位：辅导员，月薪：6000元，单位：北京印刷学院新媒体学院
class Employee():
    company = '北京印刷学院'
    def __init__(self,name,position,salary):
        self.name = name
        self.position = position
        self.salary = salary
    def show(self):
        print("姓名：{}，职位：{}，月薪：{}，单位：{}".format(self.name,self.position,self.salary,self.company))
    def annual_salary(self):
        year_salary = self.salary * 12
        print("年薪：{}".format(year_salary))

a = Employee('张三','教师',8000)
a.show()
a.annual_salary()
```

```python
#4. 请设计一个 Hero 类，用于管理革命英雄人物信息，通过编程实践铭记英雄事迹，传承革命精神。
#要求：
#（1）定义 Hero 类：
#类属性 country = "中国"
#实例属性：name（姓名）、birth（出生年份）、death（逝世年份）、deed（主要事迹）
#方法 show()：打印英雄完整信息
#方法 age()：计算英雄享年（death - birth）并返回
#方法 is_remembered()：如果英雄享年 ≥ 60 岁，返回“永垂不朽”；否则返回“英年早逝”
#（2）从以下四位英雄中任选两位，创建 Hero 对象并调用所有方法：
class Hero():
    country = '中国'
    def __init__(self,name,birth,death,deed):
        self.name = name
        self.birth = birth
        self.death = death
        self.deed = deed
    def show(self):
        print("国家：{}，姓名：{}，出生年份：{}，逝世年份：{}，主要事迹：{}".format(self.country,self.name,self.birth,self.death,self.deed))
    def age(self):
        return self.death - self.birth
    def is_remembered(self):
        if self.age() >= 60:
            return "永垂不朽"
        else:
            return "英年早逝"

a = Hero('李大钊',1889,1927,'中国共产主义运动先驱，率先在中国传播马克思主义，牵头创建中国共产党早期组织，积极投身反帝反封建革命斗争，为党的建立筑牢思想与组织根基。')
a.show()
a.is_remembered()
```
