# 集合练一练

```python
#1. 输入一个大于2的自然数，输出小于该数字的所有素数组成的集合
#（素数是指在大于1的自然数中，除了1和它本身之外不再有其他因数的自然数）

num = int(input("请输入一个大于二的数字："))
prime_set = set()

if num < 2:
    print("请确保输入数字大于2")
    
for i in range(2, num):
    is_prime = True
    for j in range(2, i):
        if i%j == 0:
            is_prime == False
            break
    if is_prime:
        prime_set.add(i)
print("小于", num, "的素数集合：", prime_set)
```

```python
#2. 编写程序，生成1000个0～100的随机整数（使用列表）
#并统计每个元素的出现次数（使用集合）

import random

lst = []
for i in range(1000):
    x = random.randint(0,100)
    lst.append(x)

s = set(lst)
for num in s:
    j = lst.count(num)
    print(num, ":", j,"次")
```

# 字典练一练

```python
#1. 有如下值集合 [11,22,33,44,55,66,7,88,99]，
#将所有大于 66 的值保存至字典的第1个key 中，
#将小于66的值保存至第2个key中。
#{‘k1':大于66的所有值，'k2':小于66所有值)

lst = [11,22,33,44,55,66,7,88,99]
d = {'k1':[],'k2':[]}

for num in lst:
    if num > 66:
        d['k1'].append(num)
    elif num < 66:
        d['k2'].append(num)
print(d)
```

```python
#2. 设计一个字典，用户输入内容作为键，然后输出字典中对应的值，
#如果用户输入的键不存在，则输出“您输入的键不存在!“。

banksystem = {"money" : 1000, "name" : "Amartareur", "age" : 19}

key = input("欢迎来到本银行系统，请问你想查询什么：")

if key in banksystem:
    print(banksystem[key])
else:
    print("您输入的键不存在")
```

```python
#3. 数字重复统计：随机生成1000个20~100 的整数，
#升序输出所有不同的数字及每个数字重复的次数。

import random

lst = []
for i in range(1000):
    x = random.randint(20,100)
    lst.append(x)

count_dict = {}

for num in lst:
    if num in count_dict:
        count_dict[num]+=1
    else:
        count_dict[num]=1

sorted_nums = sorted(count_dict.keys())

for n in sorted_nums:
    print(n, ":", count_dict[n]"次")
```
