```python
#for 循环
for n in range(100, 1, -1):
    for i in range(2, n):
        if n%i == 0:
            break
    else:#如果没有else，就会重复打印多个素数
        print(n)
```

```python
# 预设非遗项目及认定年份
intangible_heritage = {
    "昆曲": 2001,
    "剪纸": 2009,
    "端午节": 2009,
    "京剧": 2010,
    "皮影戏": 2011
}

max_queries = 3  # 最多查询次数
query_count = 0   # 当前已查询次数

while query_count < max_queries:
    query_count += 1
    print(f"\n===== 第 {query_count} 次查询 =====")
    
    # 获取用户输入的年份区间
    try:
        start_year = int(input("请输入起始年份："))
        end_year = int(input("请输入结束年份："))
    except ValueError:
        print("输入格式错误，请输入数字年份！")
        if query_count < max_queries:
            choice = input("是否继续查询？(y/n)：").strip().lower()
            if choice != 'y':
                break
        continue
    
    # 统计该年份区间内的项目
    result_list = []
    for name, year in intangible_heritage.items():
        if start_year <= year <= end_year:
            result_list.append(name)
    
    # 输出结果
    if result_list:
        print(f"{len(result_list)}个项目（{'、'.join(result_list)}）")
    else:
        print("该区间内无项目认定。")
    
    # 判断是否继续查询
    if query_count < max_queries:
        while True:
            choice = input("是否继续查询？(y/n)：").strip().lower()
            if choice in ['y', 'n']:
                break
            print("输入错误，请输入 y 或 n！")
        if choice != 'y':
            break

# 达到最大查询次数后结束
print("\n今日查询次数已用完，感谢使用！")
```
