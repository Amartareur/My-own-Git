```python
#1. 编写一个自定义异常类 ScoreError，当分数不在 0-100 范围内时抛出。编写函数 check_score(score)，如果分数有效返回 True，否则抛出 ScoreError。
#使用 try...except 捕获该异常并打印错误信息。
#测试：check_score(85)、check_score(120)、check_score(-5)。

class ScoreError(Exception):
    def __init__(self, score):
        self.score = score
        self.message = f"分数无效: {score}。分数必须在 0 到 100 之间！"
        super().__init__(self.message)

def check_score(score):
    if 0 <= score <= 100:
        return True
    else:
        raise ScoreError(score)

def test_check_score(score):
    try:
        result = check_score(score)
        print(f"检查分数 {score}: 成功 (返回值: {result})")
    except ScoreError as e:
        print(f"检查分数 {score}: 捕获到异常 -> {e}")

# --- 测试用例 ---
print("--- 开始测试 ---")
test_check_score(85)   # 有效分数
test_check_score(120)  # 超过 100
test_check_score(-5)   # 小于 0
```

```python
#2.编写一个函数 get_list_element(lst, index)，接收一个列表和一个索引，返回该索引对应的元素。
#使用 try...except 同时捕获 IndexError（索引超出范围）和 TypeError（索引不是整数）两种异常。
#测试三种情况：get_list_element([1, 2, 3], 1)、get_list_element([1, 2, 3], 5)、get_list_element([1, 2, 3], "a")

def get_list_element(lst, index):
    try:
        element = lst[index]
        return f"成功获取元素: {element}"
    except (IndexError, TypeError) as e:
        return f"发生异常: {type(e).__name__} - {e}"

print("测试 1:", get_list_element([1, 2, 3], 1))
print("测试 2:", get_list_element([1, 2, 3], 5))
print("测试 3:", get_list_element([1, 2, 3], "a"))
```

```python
#3. 编写一个函数 read_file(filename)，接收一个文件名。使用 assert 断言文件名必须是字符串类型，然后使用 try...except 捕获 FileNotFoundError。
#如果文件不存在，打印“文件未找到”；否则打印“已打开文件”并返回。
#测试两种情况：一个存在的文件和一个不存在的文件。

import os

def read_file(filename):
    assert isinstance(filename, str), "文件名必须是字符串类型"
    try:
        with open(filename, 'r') as f:
            pass
        print("已打开文件")
        return
    except FileNotFoundError:
        print("文件未找到")

existing_file = "test_exist.txt"
with open(existing_file, "w") as f:
    f.write("Hello World")

non_existing_file = "test_not_exist.txt"
print("--- 测试情况 1：文件存在 ---")
read_file(existing_file)
print("\n--- 测试情况 2：文件不存在 ---")
read_file(non_existing_file)
print("--- 测试情况 3：文件不是字符串 ---")
read_file(123)

if os.path.exists(existing_file):
    os.remove(existing_file)
```

```python
#4. 综合题：学生成绩管理系统
#       考试成绩管理中经常会出现各种输入错误——分数超出范围、输入了非数字字符等。请设计一个带完整异常处理的学生成绩管理系统。

#要求：
#（1）定义自定义异常类 ScoreError（继承 Exception）：当分数不在 0-100 范围内时抛出。

#（2）定义函数 input_score(prompt)：
#      提示用户输入一个分数
#      使用 try...except 捕获 ValueError（用户输入非数字）
#      如果输入不是数字，打印“请输入数字！”，并让用户重新输入（循环直到输入有效）
#     输入成功后，将分数转为整数返回

#（3）定义函数 add_student(students_dict, name, score)：
#      使用 try...except 捕获 ScoreError（分数不在 0-100 范围内时用 raise 抛出）
#      如果分数无效，打印错误信息并返回 False
#      如果分数有效，将学生姓名和分数存入字典，打印“添加成功”，返回 True

#（4）定义主程序 main()：
#      创建一个空字典 students = {}
#      循环让用户输入姓名和分数，调用 add_student() 添加
#      输入姓名时输入 q 退出
#      退出后打印所有学生成绩列表

class ScoreError(Exception):
    def __init__(self,score):
        self.score = score
        self.message = f"分数无效:{score}。分数必须在0-100之间"
        super().__init__(self.message)

def input_score(prompt):
    while True:
        try:
            user_input = input(prompt)
            score = int(user_input)
            return score
        except ValueError:
            print("请输入数字！")

def add_student(students_dict, name, score):
    try:
        if score < 0 or score > 100:
            raise ScoreError(score)
            
        students_dict[name] = score
        print("添加成功")
        return True
        
    except ScoreError as e:
        print(e.message)
        return False
        
def main():
    students = {}
    
    while True:
        name = input("请输入学生姓名（输入 q 退出）: ").strip()
        
        if name.lower() == 'q':
            break
            
        if not name:
            print("姓名不能为空，请重新输入。")
            continue
            
        score = input_score(f"请输入 {name} 的分数: ")
        
        add_student(students, name, score)

    print("\n====== 学生成绩列表 ======")
    if not students:
        print("暂无成绩数据。")
    else:
        for student_name, student_score in students.items():
            print(f"姓名: {student_name} \t 分数: {student_score}")
    print("==========================")

if __name__ == "__main__":
    main()
```
