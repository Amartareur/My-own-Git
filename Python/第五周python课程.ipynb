```python
 #正确的代码：
x = [1,2,1,2,2,2,1,1,1]
for i in range(len(x)-1,-1,-1): #从后往前删
   if x[i]==1:
       del x[i]
x
```

### 浅复制和深复制

```python
aList = [3, 5, 7]
aList
```

```python
bList = aList[:]
aList == bList
```

```python
aList is bList # 不是一个对象，内存地址也不一样
```

```python
bList[1] = 8
print(bList)
print(aList)
```

```python
cList = aList
id(aList) == id(cList) 
```

```python
cList[1] = 8 # 深复制就是这样的
print(cList)
print(aList)
```

```python
aList = [3, [5], 7]        #列表aList中包含可变的列表对象
bList = aList[:]           #切片
bList[1].append(6)         #调用子列表的append()方法，这个方法是原地操作的
bList
```

```python
aList = [3, [5], 7]
```

```python
import copy
bList = copy.deepcopy(aList) #深赋值，递归复制，直到遇到可哈希对象
                             #aList和bList完全独立，互相不影响
aList == bList
```

```python
bList[1].append(6)
print(bList)
print(aList)
```

# 作业

```python
s = input("请输入你想输入的10整数，中间用空格隔开：")
nums = [int(i) for i in s.split()]
new_nums = []
for num in nums:
    if num not in new_nums:
        new_nums.append(num)
print("您输入的整数是", nums)
print("删除重复后是", new_nums)
```
