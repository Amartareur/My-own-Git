```python
while True:
    userinput = input("请输入数字（1-5）查询红色精神，输入‘退出’结束程序：")

    if userinput == "退出":
        print("程序已退出，再见！")
        break
    if userinput == "1":
        print("-> 井冈山精神")
    elif userinput == "2":
        print("-> 长征精神")
    elif userinput == "3":
        print("-> 延安精神")
    elif userinput == "4":
        print("-> 西柏坡精神")
    elif userinput == "5":
        print("-> 雷锋精神")
    else:
        print("! 没有对应精神，请重新输入")
```

```python
#求100-200之间所有素数的和
s = 0

for x in range(100, 201):
    ok = True
    for i in range(2, x):
        if x % i == 0:
            ok = False
            break
    if ok:
        s = s + x

print("100~200之间所有素数的和=", s)
```
