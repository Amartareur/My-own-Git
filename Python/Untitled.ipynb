```python
#100元换成10元，5元，1元的所有组合
total = 100
count = 0

print("所有组合格式：(10元张数，5元张数，1元张数)")
print("-" * 50)

for ten in range(total // 10+1):
    rest = total - 10 * ten
    for five in range(rest // 5+1):
        one = rest - 5 * five
        count += 1
        print(f"第{count:3d}种:({ten:2d}, {five:2d}, {one:3d})")

print("-" * 50)
print(f"一共有{count}种组合")
```
