```python
# 1.编写程序，生成一个包含 20个随机整数的列表，然后对偶数下标的元素进行降序排列，奇数下标的元素不变(提示:使用切片)。
import random

random_num_20  = []
for i in range(20):
    random_num_20.append(random.randint(0,100))
print(random_num_20)
random_num_10 = sorted(random_num_20[::2],reverse=True)
print(random_num_10)
random_num_20[::2] = random_num_10
print(random_num_20)
```

```python
random_num_10 = sorted(random_num_20[1::2])
print(random_num_10)
random_num_20[::2] = random_num_10
print(random_num_20)
```
