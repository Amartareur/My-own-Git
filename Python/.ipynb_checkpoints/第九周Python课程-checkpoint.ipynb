```python
#1.编写一个程序来统计输入的英文单词的频率，按字母顺序排序后输出。
```

```python
# 获取用户输入
text = input("请输入英文单词，中间用空格隔开：")

# 全部小写 + 分割成单词列表
word_list = text.lower().split(" ")

# 统计单词频率
freq = {}
for word in word_list:
    if word in freq:
        freq[word] += 1
    else:
        freq[word] = 1

# 按单词字母顺序排序
sorted_word = sorted(freq.items())

# 输出结果
print("\n单词频率(按字母排序)：")
for word, count in sorted_word:
    print(f"{word}", "出现次数:", f"{count}", "次")
```

```python
#3. 字符串 a 为 'Hello'，字符串 b 为 '#2#Lisaend'。
#判断字符串 b 中是否含有 '#2#'，如果有，则将字符串 a 与字符串 b 中 '#2#' 与最后三个字符之间的字符串（去除可能存在的多余空格）用空格链接起来，
#然后输出 'OK' 3 次。

 #输出结果如下：

     #hello #2# end

     #okokok
```

```python
# 定义字符串
a = 'Hello'
b = '#2#Lisaend'

# 判断b是否包含 #2#
if '#2#' in b:
    # 截取最后三个字符之前的部分
    end = len(b) - 3
    
    # 取出中间字符串 + 去除空格
    mid_str = b[:end].strip()

    #去掉多余部分
    c = b.replace(mid_str, "")
    
    # a 和中间内容用空格连接输出
    res = a + " " + "#2#" + " " + c
    print(res)
    
# 输出OK三次
print("OK" * 3)
```

```python
#4. 输入一个浮点数形式字符串，也就是包含小数点（如：4.5）或采用科学计数法形式（如：9e4）的字符串。
#编写程序判断该字符串是否为浮点数形式字符串，如果是则输出为True，否则输出False。
```

```python
#获取用户输入
s = input("请输入内容：")

#定义一个变量，判断是否为浮点数
def is_float(s):
    #尝试判断浮点数
    try:
        #将属性从str转变成float
        float(s)
        # 排除纯整数！只判断浮点数，如果是纯数字返回True经过not后转为flase，而如果是浮点数返回False经过not后变为True
        return not s.isdigit()
    except:
        #如果字符串不能转换成float，直接输出False
        return False

if is_float(s):
    print("是浮点数")
else:
    print("不是浮点数")
```

```python
#5. 唐代诗人刘禹锡的《赠别》中有千古名句"他日卧龙终得雨，今朝放鹤且冲天"，这首诗表达了困境中对未来的乐观信念与豁达胸襟。
#全诗如下：鼎食华轩到眼前，拂衣高谢岂徒然。九霄路上辞朝客，四皓丛中作少年。他日卧龙终得雨，今朝放鹤且冲天。洛阳旧有衡茆在，亦拟抽身伴地仙。

#编写程序：（1）判断诗句是否以"鼎食"开头，以"地仙"结尾，分别输出 True 或 False；
        #（2）将诗句按标点拆分为七言分句，用"/"连接并输出；
        #（3）去掉所有标点符号后输出总字符数；
        #（4）用户输入一个词，程序判断该词是否在整首诗中出现，并输出出现位置及出现次数。
```

```python
poem = """鼎食华轩到眼前，拂衣高谢岂徒然。九霄路上辞朝客，四皓丛中作少年。他日卧龙终得雨，今朝放鹤且冲天。洛阳旧有衡茆在，亦拟抽身伴地仙。"""

#(1)判断是否以鼎食开头、以地仙结尾
start_res = poem.startswith("鼎食")
end_res = poem.endswith("地仙。")
print("是否以'鼎食'开头：", start_res)
print("是否以'地仙'结尾：", end_res)

# (2)按标点拆分为七言分句，用 / 连接
# 把句号、逗号统一替换，拆分
new_poem = poem.replace("，","/").replace("。","/")
# 去掉最后多余的 /
new_poem = new_poem.rstrip("/")
print("按标点拆分后：", new_poem)

#(3)去除所有标点，统计总字符数
clear_poem = poem.replace("，","").replace("。","")
count_char = len(clear_poem)
print("去除标点后总字符数：", count_char)

#(4)输入词语，判断出现次数和位置
word = input("请输入要查询的词语：")
count = poem.count(word)
pos_list = []

# 循环查找所有出现位置
start = 0
for i in range(count):
    idx = poem.find(word, start)
    pos_list.append(idx)
    start = idx + 1

print(f"词语【{word}】出现次数：{count}")
print(f"出现位置下标：{pos_list}")
```
